```python
import json
from typing import Any

import requests

SERVICE_URL = 'https://dailyfoodrecalls.com/api/recalls-latest.json'
HEADERS = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64)'}
TIMEOUT_SECONDS = 15

def fetch_recalls_list(session: requests.Session) -> list[dict[str, Any]] | None:
    print('Retrieving', SERVICE_URL)
    print('Data from <a href="https://dailyfoodrecalls.com">Daily Food Recalls</a>, licensed CC BY 4.0.')
    print('______________________________')
    print()

    try:
        response = session.get(SERVICE_URL, timeout=TIMEOUT_SECONDS)
        response.raise_for_status()
        
        data = response.json()
        return data['recalls']
    
    except requests.exceptions.RequestException as err:
        print('Request error:', err)
        return None
    
    except ValueError:
        print('Response did not contain valid JSON.')
        return None


def print_product(products: list[dict[str, Any]]) -> None:

    for product in products:
        print('Product:', product.get('product', 'N/A'))
        brand = product.get('brand') or 'N/A'
        hazard = product.get('hazard') or 'N/A'
        reason = product.get('reason') or 'N/A'
        recallInitiated = product.get('recallInitiated') or 'N/A'
        status = product.get('status') or 'N/A'
        print()
        print('Date initiated:', f'{recallInitiated}')
        print()
        print('Status:', f'{status}')
        print()
        print('Brand:', f'{brand}')
        print()
        print('Hazard:', f'{hazard}')
        print()
        print('Details:', f'{reason}')
        print()
        print('______________________________')
        print()
      
def main() -> None:
    session = requests.Session()
    session.headers.update(HEADERS)

    recalls = fetch_recalls_list(session)
    
    if not recalls:
        print('No recall data returned.')
        return

    print_product(recalls)


if __name__ == '__main__':
    main()
```
